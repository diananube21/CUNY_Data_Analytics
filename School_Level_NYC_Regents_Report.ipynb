**Cleaning** - Converting Mean Score into numeric

```python
df["Mean Score"] = pd.to_numeric(df["Mean Score"], errors="coerce")
df.head(5)

#Phyton will transform some no nemeric into NaN.
#errors="coerce", means that Phyton transfor all no muneric into NaN
```

**Choosing the subject.**

```python
geometry = df[df["Regents Exam"] == "Geometry"]
geometry.head(5)

#Analyzing only Geometry
```

```python
geometry["School Name"].unique()
```

```python
school = geometry[
    geometry["School Name"] == "Franklin Delano Roosevelt High School"
]

school
```

**Descriptves** **Statistics**

```python
school["Mean Score"].describe()
```

**We are going to compare Franklin Delano Roosevelt High School with all the schools in Brooklyn.**

```python
brooklyn = geometry[
    geometry["School Name"].str.contains("Brooklyn", case=False, na=False)
]

brooklyn["Mean Score"].describe()
```

```python
import matplotlib.pyplot as plt
plt.bar(
    ["Franklin Delano Roosevelt", "Brooklyn"],
    [school["Mean Score"].mean(), brooklyn["Mean Score"].mean()]
)

plt.ylabel("Average Mean Score")
plt.title("Geometry Regents Mean Score")
plt.show()
```

## **Conclusion**

In this analysis, I compared the Geometry Regents mean scores of Franklin Delano Roosevelt High School with the mean scores of schools across Brooklyn. Franklin Delano Roosevelt High School had an average Mean Score of 66.9, while the Brooklyn schools had an average of 62.41. This shows that, in this dataset, the school's average score was about 4.49 points higher than the Brooklyn average.
The descriptive statistics also show some differences in the range of scores. Franklin Delano Roosevelt High School had scores ranging from 53.7 to 75.1, while the Brooklyn schools had a wider range, from 33.7 to 88.9. The standard deviation for Brooklyn was also higher (12.47) compared with the school's standard deviation (6.44), indicating greater variation in the scores among Brooklyn schools. Overall, the results show that Franklin Delano Roosevelt High School had a higher average Geometry Regents score than the Brooklyn schools included in this dataset.
